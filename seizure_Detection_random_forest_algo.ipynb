```python
eeg_channels=[
    cols for cols in df_final.columns
    if cols not in ['series_id', 'p_id', 'sez']
]
```

```python
sampling_rate = 256
window_seconds = 5
samples_per_epoch = sampling_rate * window_seconds

print("Number of EEG channels:", len(eeg_channels))
print("Samples per epoch per channel:", samples_per_epoch)
print("Total samples:", len(df_final))
```

```python
epochs = []

for start in range(0, len(df_final) - samples_per_epoch + 1,
                    samples_per_epoch):

    end = start + samples_per_epoch
    window = df_final.iloc[start:end]

    # Label 1 if any sample in this window is labeled 1
    label = int(window["sez"].max())

    # Keep the EEG samples and their label
    epochs.append({
        "start_sample": start,
        "end_sample": end,
        "sez": label,
        "eeg": window[eeg_channels].to_numpy()
    })

print("Number of epochs:", len(epochs))
print("First epoch shape:", epochs[0]["eeg"].shape)
print("First epoch label:", epochs[0]["sez"])
```

```python
epoch_labels = pd.Series([epoch["sez"] for epoch in epochs])

print(epoch_labels.value_counts().sort_index())
```

```python
first_epoch = epochs[0]["eeg"]

print("Epoch shape:", first_epoch.shape)

print("Mean per channel:")
print(np.mean(first_epoch, axis=0))

print("Standard deviation per channel:")
print(np.std(first_epoch, axis=0))

print("Mean squared value per channel:")
print(np.mean(first_epoch ** 2, axis=0))
```

```python
print("Has infinite values:", np.isinf(first_epoch).any())
print("Has NaN values:", np.isnan(first_epoch).any())
print("Largest absolute value:", np.nanmax(np.abs(first_epoch)))
print("Data type:", first_epoch.dtype)
```

```python
# Convert the first epoch to a safer numeric type
first_epoch = epochs[0]["eeg"].astype(np.float64)

print("Data type:", first_epoch.dtype)
print("Standard deviation per channel:")
print(np.std(first_epoch, axis=0))
```

```python
import numpy as np
import pandas as pd

feature_rows = []

for epoch in epochs:
    # Convert EEG values to float64 to prevent overflow
    eeg = epoch["eeg"].astype(np.float64)

    # Calculate three features for each of the 22 channels
    mean_values = np.mean(eeg, axis=0)
    std_values = np.std(eeg, axis=0)
    power_values = np.mean(eeg ** 2, axis=0)

    # Store the features and the seizure label
    row = {}

    for i, channel in enumerate(eeg_channels):
        row[f"{channel}_mean"] = mean_values[i]
        row[f"{channel}_std"] = std_values[i]
        row[f"{channel}_power"] = power_values[i]

    row["sez"] = epoch["sez"]
    feature_rows.append(row)

features_df = pd.DataFrame(feature_rows)

print("Feature table shape:", features_df.shape)
print(features_df.head())
```

```python
seizure_epochs = features_df.index[features_df["sez"] == 1].tolist()

print("Seizure epoch indices:", seizure_epochs)
print("Total seizure epochs:", len(seizure_epochs))
```

```python
features_df
```

```python
from sklearn.model_selection import train_test_split

# Separate EEG features (X) from the answer labels (y)
X = features_df.drop(columns=["sez"])
y = features_df["sez"]

# Split data while keeping seizure examples in both sets
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.25,
    random_state=42,
    stratify=y
)

print("Training examples:", len(X_train))
print("Testing examples:", len(X_test))

print("\nSeizure examples in training:", int(y_train.sum()))
print("Seizure examples in testing:", int(y_test.sum()))
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report, confusion_matrix

# Create the model
model = RandomForestClassifier(
    n_estimators=100,
    class_weight="balanced",
    random_state=42
)

# Train it
model.fit(X_train, y_train)

# Predict seizure labels for the test data
y_pred = model.predict(X_test)

# Evaluate the predictions
print("Confusion matrix:")
print(confusion_matrix(y_test, y_pred, labels=[0, 1]))

print("\nClassification report:")
print(classification_report(
    y_test,
    y_pred,
    labels=[0, 1],
    target_names=["Non-seizure", "Seizure"],
    zero_division=0
))
```
